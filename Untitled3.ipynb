## Cell 4: Data Cleaning and Target Construction

### Description:
- Replace `'?'` with `NaN` for standardized missing value representation.
- Construct the target binary variable `readmitted_30d` (1 if `<30`, else 0).
- Map ICD-9 codes to clinical categories (`Circulatory`, `Respiratory`, `Digestive`, `Diabetes`, `Injury`, `Musculoskeletal`, `Genitourinary`, `Neoplasms`, `Other`).
- Record and export dataset statistics and cleaning decisions to `reports/data_cleaning_log.json`.
- Save the cleaned dataset to `data/cleaned_dataset.csv`.

```python
import numpy as np
import pandas as pd
import json
import os

# 1. Replace '?' with np.nan
df_cleaned = df.replace('?', np.nan)

# 2. Construct binary target
df_cleaned['readmitted_30d'] = (df_cleaned['readmitted'] == '<30').astype(int)

# 3. Map ICD-9 diagnostic codes to categories
def map_icd9_to_category(code):
    if pd.isna(code):
        return 'Unknown'
    code_str = str(code).strip()
    if not code_str:
        return 'Unknown'

    # Diabetes special check first (ICD-9 code 250.xx)
    if code_str.startswith('250'):
        return 'Diabetes'

    # Check for E and V codes
    if code_str.startswith('E') or code_str.startswith('V'):
        return 'Other'

    try:
        val = float(code_str)
        if 390 <= val <= 459 or val == 785:
            return 'Circulatory'
        elif 460 <= val <= 519 or val == 786:
            return 'Respiratory'
        elif 520 <= val <= 579 or val == 787:
            return 'Digestive'
        elif 800 <= val <= 999:
            return 'Injury'
        elif 710 <= val <= 739:
            return 'Musculoskeletal'
        elif 580 <= val <= 629 or val == 788:
            return 'Genitourinary'
        elif 140 <= val <= 239:
            return 'Neoplasms'
        else:
            return 'Other'
    except ValueError:
        return 'Other'

# Map the diag columns
for col in ['diag_1', 'diag_2', 'diag_3']:
    df_cleaned[f'{col}_category'] = df_cleaned[col].apply(map_icd9_to_category)

# 4. Document cleaning metadata
missing_counts = df_cleaned.isna().sum().to_dict()
cleaning_metadata = {
    "initial_shape": df.shape,
    "cleaned_shape": df_cleaned.shape,
    "missing_values_per_column_after_standardization": {k: int(v) for k, v in missing_counts.items() if v > 0},
    "target_distribution": df_cleaned['readmitted_30d'].value_counts().to_dict(),
    "diag_1_category_distribution": df_cleaned['diag_1_category'].value_counts().to_dict()
}

# Define output paths
reports_dir = '/content/drive/MyDrive/Diabetes_Readmission_ML/reports'
data_dir = '/content/drive/MyDrive/Diabetes_Readmission_ML/data'
os.makedirs(reports_dir, exist_ok=True)
os.makedirs(data_dir, exist_ok=True)

log_path = os.path.join(reports_dir, 'data_cleaning_log.json')
cleaned_csv_path = os.path.join(data_dir, 'cleaned_dataset.csv')

# 5. Save metadata to JSON
with open(log_path, 'w') as f:
    json.dump(cleaning_metadata, f, indent=4)

# 6. Save cleaned dataset
df_cleaned.to_csv(cleaned_csv_path, index=False)

print(f"Data cleaning complete.")
print(f"Cleaned dataset saved to: {cleaned_csv_path}")
print(f"Cleaning log saved to: {log_path}")
print(f"Target distribution: {cleaning_metadata['target_distribution']}")
```

## Cell 5: Patient-Grouped Train-Test Split

### Description:
- Split the cleaned dataset into an 80/20 train/test partition using `GroupShuffleSplit` based on `patient_nbr` to prevent data leakage.
- Save the partitioned sets to `data/train_split.csv` and `data/test_split.csv`.
- Verify and programmatically assert that zero patient IDs overlap between train and test partitions.
- Save a JSON split manifest describing the partition metadata to `reports/split_manifest.json`.

```python
from sklearn.model_selection import GroupShuffleSplit
import pandas as pd
import json
import os

# 1. Initialize GroupShuffleSplit with the correct keyword argument 'random_state'
gss = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)

# 2. Extract features, target, and groups
X = df_cleaned
y = df_cleaned['readmitted_30d']
groups = df_cleaned['patient_nbr']

train_idx, test_idx = next(gss.split(X, y, groups))

df_train = df_cleaned.iloc[train_idx].copy()
df_test = df_cleaned.iloc[test_idx].copy()

# 3. Verify zero patient overlap
train_patients = set(df_train['patient_nbr'])
test_patients = set(df_test['patient_nbr'])
overlap = train_patients.intersection(test_patients)

assert len(overlap) == 0, f"CRITICAL ERROR: Patient overlap detected! Number of overlapping patients: {len(overlap)}"

print("[SUCCESS] Verification passed: Zero patient overlap between partitions.")

# 4. Save splits
train_path = os.path.join(data_dir, 'train_split.csv')
test_path = os.path.join(data_dir, 'test_split.csv')

df_train.to_csv(train_path, index=False)
df_test.to_csv(test_path, index=False)

# 5. Save Split Manifest
manifest = {
    "total_records": len(df_cleaned),
    "train_records": len(df_train),
    "test_records": len(df_test),
    "unique_patients_total": df_cleaned['patient_nbr'].nunique(),
    "unique_patients_train": len(train_patients),
    "unique_patients_test": len(test_patients),
    "train_positive_prevalence": float(df_train['readmitted_30d'].mean()),
    "test_positive_prevalence": float(df_test['readmitted_30d'].mean())
}

manifest_path = os.path.join(reports_dir, 'split_manifest.json')
with open(manifest_path, 'w') as f:
    json.dump(manifest, f, indent=4)

print(f"Train split saved to: {train_path}")
print(f"Test split saved to: {test_path}")
print(f"Split manifest saved to: {manifest_path}")
print(f"Manifest summary: {json.dumps(manifest, indent=4)}")
```

## Cell 6: Leak-Free Preprocessing Pipeline

### Description:
- Identify numerical and categorical features to include in the model.
- Build a robust scikit-learn `ColumnTransformer` that imputes missing values and scales numeric variables, and imputes and one-hot encodes categorical variables.
- Ensure the preprocessing pipeline is self-contained to prevent any data leakage during cross-validation and testing.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 1. Define feature groupings
numeric_features = [
    'time_in_hospital', 'num_lab_procedures', 'num_procedures',
    'num_medications', 'number_outpatient', 'number_emergency',
    'number_inpatient', 'number_diagnoses'
]

categorical_features = [
    'race', 'gender', 'age', 'max_glu_serum', 'A1Cresult',
    'metformin', 'repaglinide', 'nateglinide', 'chlorpropamide',
    'glimepiride', 'acetohexamide', 'glipizide', 'glyburide',
    'tolbutamide', 'pioglitazone', 'rosiglitazone', 'acarbose',
    'miglitol', 'troglitazone', 'tolazamide', 'insulin',
    'glyburide-metformin', 'glipizide-metformin', 'change', 'diabetesMed',
    'diag_1_category', 'diag_2_category', 'diag_3_category'
]

# 2. Build Preprocessing Transformers
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median', add_indicator=True)),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent', add_indicator=True)),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# 3. Combine into a unified ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='drop'
)

print("[SUCCESS] Preprocessing ColumnTransformer constructed safely.")
print(f"Numerical features evaluated ({len(numeric_features)}): {numeric_features}")
print(f"Categorical features evaluated ({len(categorical_features)}): {categorical_features}")
```

## Cell 7: Patient-Grouped Five-Fold Cross-Validation

### Description:
- Use `GroupKFold` with 5 folds on the training set (`df_train`), grouping by `patient_nbr`.
- Evaluate five models: `DummyClassifier`, `LogisticRegression`, `DecisionTreeClassifier`, `RandomForestClassifier`, and `HistGradientBoostingClassifier`.
- Measure Out-of-Fold (OOF) performance using ROC AUC, Average Precision (PR AUC), and standard F1-score.
- Track and export the cross-validation performance metrics to `metrics/cv_models_performance.json`.

```python
from sklearn.model_selection import GroupKFold
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score, average_precision_score, f1_score
import numpy as np
import pandas as pd
import json
import os

# 1. Prepare target and group variables from training split
X_train = df_train
y_train = df_train['readmitted_30d'].values
groups_train = df_train['patient_nbr'].values

# 2. Define the candidate models
models = {
    "DummyClassifier": DummyClassifier(strategy="prior"),
    "LogisticRegression": LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'),
    "DecisionTreeClassifier": DecisionTreeClassifier(max_depth=10, random_state=42, class_weight='balanced'),
    "RandomForestClassifier": RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42, class_weight='balanced', n_jobs=-1),
    "HistGradientBoostingClassifier": HistGradientBoostingClassifier(max_iter=100, random_state=42, class_weight='balanced')
}

# 3. Setup GroupKFold cross-validation
gkf = GroupKFold(n_splits=5)
cv_results = {}

print("Starting Leak-Free Patient-Grouped 5-Fold Cross-Validation...\n")

for model_name, model in models.items():
    print(f"Evaluating {model_name}...")

    oof_preds_prob = np.zeros(len(df_train))
    oof_preds_class = np.zeros(len(df_train))

    for fold, (train_idx, val_idx) in enumerate(gkf.split(X_train, y_train, groups=groups_train)):
        # Split
        X_tr, y_tr = X_train.iloc[train_idx], y_train[train_idx]
        X_va, y_va = X_train.iloc[val_idx], y_train[val_idx]

        # Clone and build the full pipeline safely
        fold_pipeline = Pipeline(steps=[
            ('preprocessor', preprocessor),
            ('classifier', model)
        ])

        # Fit only on training fold
        fold_pipeline.fit(X_tr, y_tr)

        # Predict probability for the positive class
        if hasattr(fold_pipeline, "predict_proba"):
            probs = fold_pipeline.predict_proba(X_va)[:, 1]
        else:
            probs = fold_pipeline.decision_function(X_va)

        oof_preds_prob[val_idx] = probs
        oof_preds_class[val_idx] = fold_pipeline.predict(X_va)

    # Calculate Fold-Averaged / Consolidated OOF metrics
    auc_roc = roc_auc_score(y_train, oof_preds_prob)
    pr_auc = average_precision_score(y_train, oof_preds_prob)
    f1 = f1_score(y_train, oof_preds_class)

    cv_results[model_name] = {
        "oof_auc_roc": float(auc_roc),
        "oof_pr_auc": float(pr_auc),
        "oof_f1": float(f1)
    }

    print(f"  -> OOF ROC AUC: {auc_roc:.4f} | PR AUC: {pr_auc:.4f} | F1: {f1:.4f}")

# 4. Save results to metrics/
metrics_dir = '/content/drive/MyDrive/Diabetes_Readmission_ML/metrics'
os.makedirs(metrics_dir, exist_ok=True)
results_path = os.path.join(metrics_dir, 'cv_models_performance.json')

with open(results_path, 'w') as f:
    json.dump(cv_results, f, indent=4)

print(f"\n[SUCCESS] Cross-validation complete. Performance log saved to: {results_path}")
```

## Cell 8: Best Model Hyperparameter Tuning

### Description:
- Take the best model (`HistGradientBoostingClassifier`) and perform randomized hyperparameter tuning.
- Use `GroupKFold` on the training partition (`df_train`) to maintain strict separation of patients during tuning.
- Optimize hyperparameter configurations over a pre-defined grid targeting the ROC AUC metric.
- Save the cross-validation parameter tuning results to `metrics/tuning_results.json`.

```python
from sklearn.model_selection import RandomizedSearchCV
import json
import os

# 1. Define full pipeline using the chosen preprocessor and model skeleton
tuning_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', HistGradientBoostingClassifier(random_state=42, class_weight='balanced'))
])

# 2. Define search space (prefixing with 'classifier__')
param_distributions = {
    'classifier__learning_rate': [0.01, 0.05, 0.1, 0.2],
    'classifier__max_iter': [50, 100, 150],
    'classifier__max_depth': [3, 5, 8, 12],
    'classifier__min_samples_leaf': [10, 20, 50, 100]
}

# 3. Initialize RandomizedSearchCV with GroupKFold
gkf = GroupKFold(n_splits=5)
random_search = RandomizedSearchCV(
    estimator=tuning_pipeline,
    param_distributions=param_distributions,
    n_iter=8,
    scoring='roc_auc',
    cv=gkf,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# 4. Fit strictly on training split using patient groups
print("Starting Hyperparameter Optimization on training split...")
random_search.fit(X_train, y_train, groups=groups_train)

# 5. Export tuning metadata
tuning_summary = {
    "best_params": {k: (v if not isinstance(v, (np.integer, np.floating)) else v.item()) for k, v in random_search.best_params_.items()},
    "best_cv_roc_auc": float(random_search.best_score_)
}

tuning_path = os.path.join(metrics_dir, 'tuning_results.json')
with open(tuning_path, 'w') as f:
    json.dump(tuning_summary, f, indent=4)

print(f"\n[SUCCESS] Hyperparameter tuning complete.")
print(f"Best Parameters: {tuning_summary['best_params']}")
print(f"Best CV ROC AUC: {tuning_summary['best_cv_roc_auc']:.4f}")
print(f"Tuning summary saved to: {tuning_path}")
```

## Cell 9: Optimal Decision Threshold Selection

### Description:
- Reconstruct and generate Out-of-Fold (OOF) predicted probabilities using the tuned hyperparameter settings via 5-fold cross-validation on `df_train`.
- Search candidate probability thresholds from `0.0` to `1.0` (with 0.01 step resolution) to maximize the training F1-score.
- Export the optimal threshold and associated training classification performance to `metrics/optimal_threshold.json`.

```python
from sklearn.base import clone
import numpy as np
import json
import os

# 1. Retrieve best estimator pipeline
best_pipeline = random_search.best_estimator_

# 2. Extract OOF probabilities for the tuned model
gkf = GroupKFold(n_splits=5)
tuned_oof_probs = np.zeros(len(df_train))

print("Generating tuned Out-of-Fold predictions...")
for fold, (train_idx, val_idx) in enumerate(gkf.split(X_train, y_train, groups=groups_train)):
    X_tr, y_tr = X_train.iloc[train_idx], y_train[train_idx]
    X_va, y_va = X_train.iloc[val_idx], y_train[val_idx]

    fold_pipeline = clone(best_pipeline)
    fold_pipeline.fit(X_tr, y_tr)

    tuned_oof_probs[val_idx] = fold_pipeline.predict_proba(X_va)[:, 1]

# 3. Grid-search thresholds
thresholds = np.arange(0.0, 1.0, 0.01)
best_threshold = 0.5
best_f1 = 0.0

for th in thresholds:
    preds = (tuned_oof_probs >= th).astype(int)
    score = f1_score(y_train, preds)
    if score > best_f1:
        best_f1 = score
        best_threshold = th

# 4. Save optimal threshold parameters
threshold_metadata = {
    "optimal_threshold": float(best_threshold),
    "training_oof_max_f1": float(best_f1)
}

threshold_path = os.path.join(metrics_dir, 'optimal_threshold.json')
with open(threshold_path, 'w') as f:
    json.dump(threshold_metadata, f, indent=4)

print(f"\n[SUCCESS] Threshold selection complete.")
print(f"Optimal Decision Threshold: {best_threshold:.4f}")
print(f"Maximum OOF Training F1-Score: {best_f1:.4f}")
print(f"Saved to: {threshold_path}")
```

## Cell 10: Final Model Evaluation on the Untouched Test Set

### Description:
- Fit the finalized, tuned `HistGradientBoostingClassifier` on the complete training set (`df_train`).
- Generate prediction probabilities and binary classifications (applying the optimal threshold of 0.5400) on the untouched test set (`df_test`).
- Evaluate generalization performance across ROC AUC, PR AUC, standard F1-score, precision, recall, and accuracy.
- Export these test set metrics to `metrics/final_test_performance.json`.

```python
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score
import joblib
import json
import os

# 1. Prepare entire train and test feature sets
X_train = df_train
y_train = df_train['readmitted_30d'].values
X_test = df_test
y_test = df_test['readmitted_30d'].values

# 2. Fit the tuned pipeline on the entire training dataset
print("Fitting finalized model on the complete training split...")
final_pipeline = random_search.best_estimator_
final_pipeline.fit(X_train, y_train)

# 3. Predict on the untouched test set
print("Generating test set predictions...")
test_probs = final_pipeline.predict_proba(X_test)[:, 1]

# Retrieve optimal threshold
opt_threshold = threshold_metadata["optimal_threshold"]
test_preds = (test_probs >= opt_threshold).astype(int)

# 4. Calculate final performance metrics
test_auc_roc = roc_auc_score(y_test, test_probs)
test_pr_auc = average_precision_score(y_test, test_probs)
test_accuracy = accuracy_score(y_test, test_preds)
test_precision = precision_score(y_test, test_preds)
test_recall = recall_score(y_test, test_preds)
test_f1 = f1_score(y_test, test_preds)

final_metrics = {
    "test_auc_roc": float(test_auc_roc),
    "test_pr_auc": float(test_pr_auc),
    "test_accuracy": float(test_accuracy),
    "test_precision": float(test_precision),
    "test_recall": float(test_recall),
    "test_f1": float(test_f1)
}

# 5. Export metrics and save model
models_dir = '/content/drive/MyDrive/Diabetes_Readmission_ML/models'
os.makedirs(models_dir, exist_ok=True)
model_save_path = os.path.join(models_dir, 'final_model.joblib')
joblib.dump(final_pipeline, model_save_path)

final_metrics_path = os.path.join(metrics_dir, 'final_test_performance.json')
with open(final_metrics_path, 'w') as f:
    json.dump(final_metrics, f, indent=4)

print(f"\n[SUCCESS] Final Evaluation Complete.")
print(f"Model artifact saved to: {model_save_path}")
print(f"Final test metrics saved to: {final_metrics_path}")
print(f"Test ROC AUC: {test_auc_roc:.4f} | PR AUC: {test_pr_auc:.4f} | F1-Score: {test_f1:.4f}")
```

## Cell 11: Export Test Predictions

### Description:

## Cell 14: System-wide Pipeline Verification

### Description:
- Set up a systematic check across all generated dataset splits, serialization models, evaluation metrics, prediction matrices, and reports.
- Programmatically verify that all folders and critical files exist, have non-zero sizes, and adhere to structural requirements.
- Confirm zero patient overlap between train/test partitions.
- Save the final verification log structure as `reports/phase13_verification.json` to establish absolute clinical pipeline sanity.

```python
import os
import json
import joblib
import pandas as pd

# 1. Setup list of target files and validation checks
project_root = '/content/drive/MyDrive/Diabetes_Readmission_ML'
verification_log = {
    "folder_checks": {},
    "file_existence_checks": {},
    "data_integrity_checks": {},
    "model_check": {}
}

# Check all expected directory structures
folders_to_check = ['data', 'models', 'reports', 'figures', 'metrics', 'predictions', 'logs']
for folder in folders_to_check:
    folder_path = os.path.join(project_root, folder)
    verification_log["folder_checks"][folder] = os.path.exists(folder_path)

# Check all required generated artifact files
files_to_check = {
    "cleaned_dataset": "data/cleaned_dataset.csv",
    "train_split": "data/train_split.csv",
    "test_split": "data/test_split.csv",
    "final_model_weights": "models/final_model.joblib",
    "data_cleaning_log": "reports/data_cleaning_log.json",
    "split_manifest": "reports/split_manifest.json",
    "cv_performance": "metrics/cv_models_performance.json",
    "tuning_results": "metrics/tuning_results.json",
    "optimal_threshold": "metrics/optimal_threshold.json",
    "final_performance": "metrics/final_test_performance.json",
    "test_predictions": "predictions/test_predictions.csv",
    "roc_curve": "figures/roc_curve.png",
    "pr_curve": "figures/pr_curve.png",
    "final_report": "reports/final_report.md"
}

all_files_exist = True
for key, relative_path in files_to_check.items():
    full_path = os.path.join(project_root, relative_path)
    exists = os.path.exists(full_path)
    file_size = os.path.getsize(full_path) if exists else 0
    verification_log["file_existence_checks"][key] = {
        "exists": exists,
        "file_size_bytes": file_size
    }
    if not exists or file_size == 0:
        all_files_exist = False

# 2. Perform deep programmatic patient-overlap and data shape validations
try:
    # Load datasets
    train_df = pd.read_csv(os.path.join(project_root, "data/train_split.csv"))
    test_df = pd.read_csv(os.path.join(project_root, "data/test_split.csv"))
    pred_df = pd.read_csv(os.path.join(project_root, "predictions/test_predictions.csv"))

    # Check patient overlap
    train_patients = set(train_df['patient_nbr'])
    test_patients = set(test_df['patient_nbr'])
    overlap = train_patients.intersection(test_patients)

    verification_log["data_integrity_checks"] = {
        "train_split_records": len(train_df),
        "test_split_records": len(test_df),
        "predictions_records": len(pred_df),
        "predictions_match_test_shape": len(pred_df) == len(test_df),
        "overlapping_patient_ids_count": len(overlap),
        "leakage_free_assertion_passed": len(overlap) == 0
    }
except Exception as e:
    verification_log["data_integrity_checks"] = {
        "status": "failed",
        "error": str(e)
    }

# 3. Model validation via deserialization
try:
    model_path = os.path.join(project_root, "models/final_model.joblib")
    loaded_model = joblib.load(model_path)
    verification_log["model_check"] = {
        "load_successful": True,
        "steps_in_pipeline": [name for name, _ in loaded_model.steps]
    }
except Exception as e:
    verification_log["model_check"] = {
        "load_successful": False,
        "error": str(e)
    }

# 4. Save the verification manifest output
verification_log["pipeline_passed_all_checks"] = (
    all_files_exist and
    verification_log["data_integrity_checks"].get("leakage_free_assertion_passed", False) and
    verification_log["model_check"].get("load_successful", False)
)

verification_json_path = os.path.join(project_root, "reports/phase13_verification.json")
with open(verification_json_path, 'w') as f:
    json.dump(verification_log, f, indent=4)

print("=== Pipeline Verification Manifest Results ===")
print(json.dumps(verification_log, indent=4))
print(f"\nVerification manifest successfully written to: {verification_json_path}")
```

## Feature Importance Analysis

### Description:
- Load the final serialized pipeline (`models/final_model.joblib`).
- Compute permutation feature importances on the training set to identify which features drive early readmissions.
- Plot the top 15 most important features and save the plot as `figures/feature_importance.png`.

```python
import os
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.inspection import permutation_importance

# 1. Paths
project_root = '/content/drive/MyDrive/Diabetes_Readmission_ML'
model_path = os.path.join(project_root, 'models/final_model.joblib')
figures_dir = os.path.join(project_root, 'figures')

# 2. Load model pipeline and training data
final_pipeline = joblib.load(model_path)
train_df = pd.read_csv(os.path.join(project_root, 'data/train_split.csv'))

y_train = train_df['readmitted_30d'].values
# We drop the identifiers and target to feed only features to the preprocessor
X_train = train_df.drop(columns=['encounter_id', 'patient_nbr', 'readmitted', 'readmitted_30d'])

# 3. Compute Permutation Feature Importance
print("Computing permutation feature importance on training split...")
result = permutation_importance(
    final_pipeline, X_train, y_train, n_repeats=5, random_state=42, n_jobs=-1
)

# 4. Formulate importance dataframe
importances = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': result.importances_mean,
    'Std': result.importances_std
}).sort_values(by='Importance', ascending=False)

# 5. Plot top 15 features
top_features = importances.head(15)

plt.figure(figsize=(10, 6))
plt.barh(y=top_features['Feature'][::-1], width=top_features['Importance'][::-1],
         xerr=top_features['Std'][::-1], color='teal', edgecolor='black', alpha=0.8)
plt.xlabel('Mean Decrease in Accuracy / Score')
plt.title('Top 15 Most Important Features (Permutation Importance)')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()

# Save visual representation
importance_plot_path = os.path.join(figures_dir, 'feature_importance.png')
plt.savefig(importance_plot_path, dpi=300, bbox_inches='tight')
plt.show()

print(f"[SUCCESS] Feature importance plot saved to: {importance_plot_path}")
display(importances.head(15))
```
